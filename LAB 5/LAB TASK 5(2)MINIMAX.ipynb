```python
#MINIMAX

import math
visited = 0
scores = [1, 2, 3, 4, 5, 6, 7, 8]
depth = int(math.log2(len(scores)))
def minimax(level, position, max_player, depth):
    global visited
    if level == depth:
        visited += 1
        return scores[position]

    if max_player:
        left = minimax(level + 1, position * 2, False, depth)
        right = minimax(level + 1, position * 2 + 1, False, depth)
        return max(left, right)

    else:
        left = minimax(level + 1, position * 2, True, depth)
        right = minimax(level + 1, position * 2 + 1, True, depth)
        return min(left, right)
result = minimax(0, 0, True, depth)

print("Best Value:", result)
print("Visited Nodes:", visited)
```
