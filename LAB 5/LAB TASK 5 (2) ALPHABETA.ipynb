```python

import math

visited = 0

def alpha_beta(level, position, max_player, values, limit, a, b):
    global visited

    if level == limit:
        visited += 1
        return values[position]

    if max_player:
        best = float("-inf")

        for j in range(2):
            result = alpha_beta(
                level + 1,
                position * 2 + j,
                False,
                values,
                limit,
                a,
                b
            )

            best = max(best, result)
            a = max(a, best)

            if a >= b:
                break

        return best

    else:
        best = float("inf")

        for j in range(2):
            result = alpha_beta(
                level + 1,
                position * 2 + j,
                True,
                values,
                limit,
                a,
                b
            )

            best = min(best, result)
            b = min(b, best)

            if a >= b:
                break

        return best


values = [3, 5, 2, 9, 3, 5, 2, 9]
limit = int(math.log2(len(values)))

answer = alpha_beta(
    0, 0, True, values, limit,
    float("-inf"), float("inf")
)

print("Best Value:", answer)
print("Visited Nodes:", visited)
```
