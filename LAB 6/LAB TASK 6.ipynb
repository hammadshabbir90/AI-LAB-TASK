```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


texts = [
    "Python is easy!",
    "Where is my account?",
    "student@gmail.com",
    "SE2026 version 3.5",
    "Hello world."
]


def split_words(text):
    return text.split()


def find_words(text):
    return re.findall(r"\b\w+\b", text)


for count, sentence in enumerate(texts, 1):

    first = split_words(sentence)
    second = find_words(sentence)

    print("Text Number:", count)
    print("Sentence:", sentence)
    print("Split Words:", first)
    print("Total Words:", len(first))
    print("Found Words:", second)
    print("Total Found:", len(second))
    print("------------------------------")


print("Empty Check:", find_words("") == [])


faqs = [
    {
        "question": "How can I change my username?",
        "answer": "Open your account settings and select the edit option to change your username.",
        "source": "Account Guide"
    },
    {
        "question": "Where can I check my exam schedule?",
        "answer": "Open the student portal and visit the examination section.",
        "source": "Student Portal Guide"
    },
    {
        "question": "How can I download my result?",
        "answer": "Visit the result section, select your semester and download the result.",
        "source": "Result Guide"
    },
    {
        "question": "How do I update my profile?",
        "answer": "Go to your profile settings and update the required information.",
        "source": "Profile Guide"
    },
    {
        "question": "How can I contact the IT department?",
        "answer": "Visit the help section of the student portal to find contact information.",
        "source": "IT Support Guide"
    }
]


question_list = []

for data in faqs:
    question_list.append(data["question"])


model = TfidfVectorizer(stop_words="english")

question_data = model.fit_transform(question_list)


def get_answer(user_text):

    user_text = user_text.strip()

    if user_text == "":
        return "Please write something."

    for data in faqs:

        if user_text.lower() == data["question"].lower():
            return "Answer: " + data["answer"] + "\nSource: " + data["source"]

    new_question = model.transform([user_text])

    result = cosine_similarity(new_question, question_data)[0]

    highest = 0
    selected = 0

    for i in range(len(result)):

        if result[i] > highest:
            highest = result[i]
            selected = i

    if highest < 0.15:
        return "Sorry, no matching question was found."

    return (
        "Question: " + faqs[selected]["question"]
        + "\nScore: " + str(round(highest, 3))
        + "\nAnswer: " + faqs[selected]["answer"]
        + "\nSource: " + faqs[selected]["source"]
    )


print("Chatbot Started")


while True:

    message = input("\nEnter Question: ")

    if message.lower() == "exit":
        print("Program Closed")
        break

    reply = get_answer(message)

    print(reply)
```
