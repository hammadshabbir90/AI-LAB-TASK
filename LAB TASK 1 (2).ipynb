```python

def checkApplication(score, documents, projects, days):
    problems = []

    if score < 70:
        problems.append("Low score")

    if documents == False:
        problems.append("Missing documents")

    if projects < 70:
        problems.append("Low projects")

    if days > 7:
        problems.append("Late application")

    if problems == []:
        return "Review", "All requirements are complete"
    else:
        return "Hold", ", ".join(problems)


applications = [
    ("Student A", 85, True, 3, 4, "Review"),
    ("Student B", 90, True, 1, 3, "Hold"),
    ("Student C", 78, True, 4, 9, "Hold"),
    ("Student D", 70, True, 2, 7, "Review"),
]

for name, score, documents, projects, days, expected in applications:

    actual, reason = checkApplication(score, documents, projects, days)

    print("Student:", name)
    print("Score:", score)
    print("Documents:", documents)
    print("Projects:", projects)
    print("Days:", days)
    print("Expected:", expected)
    print("Actual:", actual)
    print("Reason:", reason)
    print("Test passed:", expected == actual)
```
