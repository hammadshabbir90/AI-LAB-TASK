```python

def chooseMode(temp, occupied, oldMode):

    if not occupied:
        newMode = "ECO"
    elif temp > 26:
        newMode = "COOL"
    elif temp < 20:
        newMode = "WARM"
    else:
        newMode = "IDLE"

    if newMode == oldMode:
        return newMode, False

    return newMode, True


testData = [
    (30, True),
    (30, True),
    (26, True),
    (20, True),
    (18, True),
    (18, False)
]

oldMode = None

for temp, occupied in testData:

    newMode, sendCommand = chooseMode(temp, occupied, oldMode)

    print("Previous Mode:", oldMode)
    print("Room Temperature:", temp)
    print("Room Status:", occupied)
    print("Selected Mode:", newMode)

    if sendCommand:
        print("Command Sent")
    else:
        print("No New Command")

    print()
    oldMode = newMode
```
