```python
GRP = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": ["G"],
    "G": []
}
# BFS
visited_bfs = []
queue = []

def bfs(start):
    queue.append(start)
    visited_bfs.append(start)

    while queue:
        print("Queue:", queue, "Visited:", visited_bfs)
        current = queue.pop(0)

        for nextNode in GRP[current]:
            if nextNode not in visited_bfs:
                visited_bfs.append(nextNode)
                queue.append(nextNode)

bfs("A")

# DFS
visited_dfs = []
stack = []

def dfs(start):
    stack.append(start)

    while stack:
        print("Stack:", stack, "Visited:", visited_dfs)
        current = stack.pop()

        if current not in visited_dfs:
            visited_dfs.append(current)
            for nextNode in reversed(GRP[current]):
                if nextNode not in visited_dfs:
                    stack.append(nextNode)

dfs("A")
```
