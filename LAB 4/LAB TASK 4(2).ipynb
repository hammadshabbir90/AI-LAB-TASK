```python

# Graph
graph = {
    "A": {"B": 2, "C": 4},
    "B": {"D": 2},
    "C": {"E": 3},
    "D": {"C": 2, "E": 4},
    "E": {"G": 7},
    "G": {}
}

# Heuristic values
hValues = {
    "A": 20, "B": 4, "C": 3,
    "D": 5, "E": 10, "G": 0
}


def searchPath(hValues, searchType):
    paths = [("A", 0, ["A"])]
    bestCost = {"A": 0}
    visited = []

    print("\n---", searchType, "---")

    while paths:
        paths.sort(key=lambda x: x[1] + hValues[x[0]])

        current, cost, route = paths.pop(0)

        if cost != bestCost[current]:
            continue

        visited.append(current)

        print("Current Node:", current,
              "| Cost:", cost,
              "| Priority:", cost + hValues[current])

        if current == "G":
            print("Final Path:", " -> ".join(route))
            print("Total Cost:", cost)
            print("Visited Nodes:", visited)
            print("Total Checked:", len(visited))
            return

        for nextNode, distance in graph[current].items():
            newCost = cost + distance

            if newCost < bestCost.get(nextNode, float("inf")):
                bestCost[nextNode] = newCost
                paths.append(
                    (nextNode, newCost, route + [nextNode])
                )


searchPath(hValues, "A Star Search")

zeroValues = {node: 0 for node in graph}
searchPath(zeroValues, "Uniform Cost Search")

newValues = hValues.copy()
newValues["C"] = 15

searchPath(newValues, "Modified A Star Search")
```
